```python
import pandas as pd
import matplotlib.pyplot as plt

# -----------------------------
# Load data
# -----------------------------
df = pd.read_csv("Figure 5 Processed - Sheet1.csv")

# Convert date column
df["Rebalance Date"] = pd.to_datetime(df["Rebalance Date"])

# Set date as index (optional but clean)
df = df.set_index("Rebalance Date")

# -----------------------------
# Preview data
# -----------------------------
display(df)
print(df.describe())

# -----------------------------
# 1. Annual Returns Over Time
# -----------------------------
plt.figure()
plt.plot(df.index, df["Vol Target Ann Ret"], label="Vol Target")
plt.plot(df.index, df["Sharpe Constrained Ann Ret"], label="Sharpe Constrained")
plt.title("Annual Returns Over Time", fontsize=10)
plt.xlabel("Rebalance Date", fontsize=8)
plt.ylabel("Annual Return", fontsize=8)
plt.legend(fontsize=8)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.show()

# -----------------------------
# 2. Annual Volatility Over Time
# -----------------------------
plt.figure()
plt.plot(df.index, df["Vol Target Ann Vol"], label="Vol Target")
plt.plot(df.index, df["Sharpe Constrained Ann Vol"], label="Sharpe Constrained")
plt.title("Annual Volatility Over Time", fontsize=10)
plt.xlabel("Rebalance Date", fontsize=8)
plt.ylabel("Annual Volatility", fontsize=8)
plt.legend(fontsize=8)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.show()

# -----------------------------
# 3. Sharpe Ratios Over Time
# -----------------------------
plt.figure()
plt.plot(df.index, df["Vol Target Sharpe"], label="Vol Target")
plt.plot(df.index, df["Sharpe Constrained Sharpe"], label="Sharpe Constrained")
plt.title("Sharpe Ratios Over Time", fontsize=10)
plt.xlabel("Rebalance Date", fontsize=8)
plt.ylabel("Sharpe Ratio", fontsize=8)
plt.legend(fontsize=8)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.show()

# -----------------------------
# 4. Maximum Drawdown Over Time
# -----------------------------
plt.figure()
plt.plot(df.index, df["Vol Target MaxDD"], label="Vol Target")
plt.plot(df.index, df["Sharpe Constrained MaxDD"], label="Sharpe Constrained")
plt.title("Maximum Drawdown Over Time", fontsize=10)
plt.xlabel("Rebalance Date", fontsize=8)
plt.ylabel("Max Drawdown", fontsize=8)
plt.legend(fontsize=8)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.show()

# -----------------------------
# 5. Side-by-Side Comparison (Returns vs Risk)
# -----------------------------
plt.figure()
plt.scatter(df["Vol Target Ann Vol"], df["Vol Target Ann Ret"], label="Vol Target")
plt.scatter(df["Sharpe Constrained Ann Vol"], df["Sharpe Constrained Ann Ret"], label="Sharpe Constrained")
plt.title("Risk vs Return", fontsize=10)
plt.xlabel("Annual Volatility", fontsize=8)
plt.ylabel("Annual Return", fontsize=8)
plt.legend(fontsize=8)
plt.xticks(fontsize=8)
plt.yticks(fontsize=8)
plt.show()
```
